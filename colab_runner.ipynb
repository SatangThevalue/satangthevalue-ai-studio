This notebook downloads the latest project code from GitHub, mounts your Google Drive for permanent data storage, and starts the FastAPI + Gradio server.

```python
# 1. Install required packages (uv, ffmpeg)
!apt-get update && apt-get install -y ffmpeg
!pip install uv
```

```python
# 2. Clone OR Update Code from GitHub (Stateless Code)
# The code lives temporarily in Colab so you always get the latest updates.
import os

REPO_URL = "https://github.com/SatangThevalue/satangthevalue-ai-studio.git"
REPO_DIR = "/content/satangthevalue-ai-studio"

if not os.path.exists(REPO_DIR):
    print("Cloning repository for the first time...")
    !git clone {REPO_URL} {REPO_DIR}
else:
    print("Repository exists. Pulling latest updates...")
    os.chdir(REPO_DIR)
    !git pull

# Move into the project directory
os.chdir(REPO_DIR)
```

```python
# 3. Install dependencies
!uv pip install -r pyproject.toml --system
```

```python
# 4. Install localtunnel
!npm install -g localtunnel
```

```python
# 5. Define Custom Workspace & Mount Google Drive (Stateful Data)
from google.colab import drive
import os
drive.mount('/content/drive')

# 🔴 YOU CAN CHANGE YOUR WORKSPACE FOLDER NAME HERE 🔴
WORKSPACE_NAME = "satangthevalue-ai-studio"
workspace_dir = f"/content/drive/MyDrive/{WORKSPACE_NAME}"
os.makedirs(workspace_dir, exist_ok=True)

# Set environment variables so the app knows where to save files
os.environ['APP_WORKSPACE_DIR'] = workspace_dir
os.environ['HF_HOME'] = f"{workspace_dir}/models_cache"

print(f"✅ Workspace is set to: {workspace_dir}")
print(f"✅ All your outputs, databases, and AI models will be safely stored here!")
```

```python
# 6. Run the FastAPI server in the background and expose via localtunnel
import subprocess
import time
import os

env = os.environ.copy()
subprocess.Popen(["uvicorn", "src.main:app", "--host", "0.0.0.0", "--port", "8000"], env=env)
time.sleep(3)

print("Starting localtunnel... Click the link below!")
!curl -s ipv4.icanhazip.com
!lt --port 8000
```
