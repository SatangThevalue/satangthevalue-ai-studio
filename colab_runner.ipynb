This notebook uses Google Drive to cache packages and models, making your 2nd run significantly faster.

```python
# 1. Mount Google Drive FIRST (Crucial for caching speed)
from google.colab import drive
import os
drive.mount('/content/drive')

WORKSPACE_NAME = "satangthevalue-ai-studio"
workspace_dir = f"/content/drive/MyDrive/{WORKSPACE_NAME}"
os.makedirs(workspace_dir, exist_ok=True)

# Set environment variables for extreme caching
os.environ['APP_WORKSPACE_DIR'] = workspace_dir
os.environ['HF_HOME'] = f"{workspace_dir}/models_cache"
os.environ['UV_CACHE_DIR'] = f"{workspace_dir}/.uv_cache" # Cache Python packages to Drive!

print(f"✅ Workspace & Cache mapped to: {workspace_dir}")
```
